3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
# ============= ЗАДАНИЕ 2.3. Опыт работы =============

import re

# Функция для извлечения опыта работы в месяцах
def get_experience(exp_str):
    """
    Извлекает общий опыт работы в месяцах из строки.
    
    Форматы:
    - 'Опыт работы 16 лет 10 месяцев ...'
    - 'Опыт работы 4 года ...'
    - 'Опыт работы 11 месяцев ...'
    - 'Не указано' → NaN
    """
    # Обработка пропусков
    if pd.isna(exp_str):
        return np.nan
    
    # Если указано "Не указано"
    if exp_str.strip() == 'Не указано':
        return np.nan
    
    # Ищем "X лет" и "Y месяцев" в начале строки
    years = 0
    months = 0
    
    # Извлекаем годы
    years_match = re.search(r'(\d+)\s+(?:год|года|лет)', exp_str)
    if years_match:
        years = int(years_match.group(1))
    
    # Извлекаем месяцы
    months_match = re.search(r'(\d+)\s+(?:месяц|месяца|месяцев)', exp_str)
    if months_match:
        months = int(months_match.group(1))
    
    # Общий опыт в месяцах
    return years * 12 + months


# Создаём новый признак
df['Опыт работы (месяц)'] = df['Опыт работы'].apply(get_experience)

# Проверяем
print("Статистика по опыту работы (в месяцах):")
print(df['Опыт работы (месяц)'].describe())

print(f"\nПропусков: {df['Опыт работы (месяц)'].isna().sum()}")

# Удаляем исходный признак
df = df.drop(columns=['Опыт работы'])
print(f"\nСтолбец 'Опыт работы' удалён")
print(f"Всего столбцов: {df.shape[1]}")
print(f"Столбцы: {df.columns.tolist()}")
```

```python
# ============= ЗАДАНИЕ 2.4. Город, переезд, командировки =============

# Список городов-миллионников (из задания)
million_cities = ['Новосибирск', 'Екатеринбург', 'Нижний Новгород', 
                  'Казань', 'Челябинск', 'Омск', 'Самара', 
                  'Ростов-на-Дону', 'Уфа', 'Красноярск', 
                  'Пермь', 'Воронеж', 'Волгоград']


# Функция для извлечения города
def get_city(row):
    """Извлекает город и определяет категорию."""
    # Город — первый элемент до ' , '
    city = row.split(' , ')[0].strip()
    
    if city == 'Москва':
        return 'Москва'
    elif city == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    elif city in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'


# Функция для готовности к переезду
def get_relocation(row):
    """Определяет готовность к переезду."""
    if 'не готов к переезду' in row or 'не готова к переезду' in row:
        return False
    elif 'готов к переезду' in row or 'готова к переезду' in row:
        return True
    elif 'хочу переехать' in row:
        return True
    return False


# Функция для готовности к командировкам
def get_business_trips(row):
    """Определяет готовность к командировкам."""
    if 'не готов к командировкам' in row or 'не готова к командировкам' in row:
        return False
    elif 'готов к командировкам' in row or 'готова к командировкам' in row:
        return True
    elif 'готова к редким командировкам' in row or 'готов к редким командировкам' in row:
        return True
    return False


# Создаём признаки
df['Город'] = df['Город, переезд, командировки'].apply(get_city)
df['Готовность к переезду'] = df['Город, переезд, командировки'].apply(get_relocation)
df['Готовность к командировкам'] = df['Город, переезд, командировки'].apply(get_business_trips)

# Проверяем
print("Распределение по городам:")
print(df['Город'].value_counts())

print("\nГотовность к переезду:")
print(df['Готовность к переезду'].value_counts())

print("\nГотовность к командировкам:")
print(df['Готовность к командировкам'].value_counts())

# Удаляем исходный признак
df = df.drop(columns=['Город, переезд, командировки'])
print(f"\nСтолбец 'Город, переезд, командировки' удалён")
print(f"Всего столбцов: {df.shape[1]}")
print(f"Столбцы: {df.columns.tolist()}")
```

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>
Выполните данное преобразование для признаков "Занятость" и "График", ответьте на контрольные вопросы, после чего удалите их из таблицы

```python
# ============= ЗАДАНИЕ 2.5. Занятость и График (One-Hot Encoding) =============

# Список категорий занятости
employment_types = ['полная занятость', 'частичная занятость', 
                    'проектная работа', 'волонтерство', 'стажировка']

# Список категорий графика
schedule_types = ['полный день', 'сменный график', 'гибкий график', 
                  'удаленная работа', 'вахтовый метод']


# Создаём бинарные признаки для "Занятость"
for emp in employment_types:
    col_name = f'Занятость_{emp}'
    df[col_name] = df['Занятость'].str.contains(emp, case=False, na=False)


# Создаём бинарные признаки для "График"
for sch in schedule_types:
    col_name = f'График_{sch}'
    df[col_name] = df['График'].str.contains(sch, case=False, na=False)


# Проверяем
print("Признаки по занятости:")
for emp in employment_types:
    col_name = f'Занятость_{emp}'
    print(f"  {col_name}: {df[col_name].sum()}")

print("\nПризнаки по графику:")
for sch in schedule_types:
    col_name = f'График_{sch}'
    print(f"  {col_name}: {df[col_name].sum()}")

# Удаляем исходные признаки
df = df.drop(columns=['Занятость', 'График'])
print(f"\nСтолбцы 'Занятость' и 'График' удалены")
print(f"Всего столбцов: {df.shape[1]}")
print(f"Столбцы: {df.columns.tolist()}")
```

**График (5):** полный день (41 716), гибкий график (15 584), удалённая работа (15 022), сменный график (12 725), вахтовый метод (3 084).

Мы уже сделали выгрузку курсов валют, которые встречаются в наших данных за период с 29.12.2017 по 05.12.2019. Скачать ее вы можете **на платформе**

Создайте новый DataFrame из полученного файла. В полученной таблице нас будут интересовать столбцы:
* "currency" - наименование валюты в ISO кодировке,
* "date" - дата, 
* "proportion" - пропорция, 
* "close" - цена закрытия (последний зафиксированный курс валюты на указанный день).


Перед вами таблица соответствия наименований иностранных валют в наших данных и их общепринятых сокращений, которые представлены в нашем файле с курсами валют. Пропорция - это число, за сколько единиц валюты указан курс в таблице с курсами. Например, для казахстанского тенге курс на 20.08.2019 составляет 17.197 руб. за 100 тенге, тогда итоговый курс равен - 17.197 / 100 = 0.17197 руб за 1 тенге.

Итак, давайте обсудим возможный алгоритм преобразования: 
1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
# ============= ЗАДАНИЕ 2.6. ЗП в рублях =============

import re

# 1. Приводим даты к datetime
currency['date'] = pd.to_datetime(currency['date'], format='%d/%m/%y').dt.date
df['Обновление резюме'] = pd.to_datetime(df['Обновление резюме'], dayfirst=True)
df['date'] = df['Обновление резюме'].dt.date

# 2. Словарь валют (ВАЖНО: 'бел.руб.' до 'руб.'!)
currency_map = {
    'бел.руб.': 'BYN',
    'руб.': 'RUB',
    'грн.': 'UAH',
    'сум': 'UZS'
}

# 3. Функция извлечения суммы
def extract_salary(zp_str):
    """Извлекает сумму из строки ЗП."""
    if pd.isna(zp_str):
        return np.nan
    match = re.search(r'([\d\s]+)', str(zp_str))
    if match:
        return float(match.group(1).replace(' ', ''))
    return np.nan

# 4. Функция извлечения валюты
def extract_currency(zp_str):
    """Извлекает валюту из строки ЗП."""
    if pd.isna(zp_str):
        return np.nan
    for key, val in currency_map.items():
        if key in str(zp_str):
            return val
    return np.nan

# 5. Создаём промежуточные столбцы
df['salary_amount'] = df['ЗП'].apply(extract_salary)
df['salary_currency'] = df['ЗП'].apply(extract_currency)

print("Валюты после парсинга:")
print(df['salary_currency'].value_counts())

# 6. Присоединяем курсы
df = df.merge(
    currency[['currency', 'date', 'close', 'proportion']],
    left_on=['salary_currency', 'date'],
    right_on=['currency', 'date'],
    how='left'
)

# 7. Обработка рубля (курс = 1)
rub_mask = df['salary_currency'] == 'RUB'
df.loc[rub_mask, 'close'] = 1
df.loc[rub_mask, 'proportion'] = 1

# 8. Считаем ЗП в рублях
df['ЗП (руб)'] = df['salary_amount'] * df['close'] / df['proportion']

# 9. Проверка
print(f"\nСтрок после merge: {df.shape[0]}")
print(f"Пропусков в ЗП (руб): {df['ЗП (руб)'].isna().sum()}")
print(f"\nСтатистика ЗП (руб):")
print(df['ЗП (руб)'].describe())

# 10. Удаляем промежуточные столбцы
df = df.drop(columns=['ЗП', 'date', 'currency', 'close', 'proportion',
                       'salary_amount', 'salary_currency', 'Обновление резюме'],
             errors='ignore')

print(f"\nВсего столбцов: {df.shape[1]}")
print(f"Столбцы: {df.columns.tolist()}")
```

Попробуйте построить распределение признака в **логарифмическом масштабе**. Добавьте к графику линии, отображающие **среднее и границы интервала метода трех сигм**. Напомним, сделать это можно с помощью метода axvline. Например, для построение линии среднего будет иметь вид:

`histplot.axvline(log_age.mean(), color='k', lw=2)`

В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
# ============= ЗАДАНИЕ 4.6. Возраст: z-отклонение =============

import numpy as np

print("Строк до очистки:", df.shape[0])

# Логарифмируем возраст
log_age = np.log(df['Возраст'])

# Считаем среднее и std логарифма
mean_log = log_age.mean()
std_log = log_age.std()

print(f"Среднее log(возраст): {mean_log:.4f}")
print(f"Std log(возраст): {std_log:.4f}")

# Гистограмма в логарифмическом масштабе
fig, ax = plt.subplots(figsize=(12, 6))
sns.histplot(log_age, bins=50, kde=True, color='teal', ax=ax)
ax.axvline(mean_log, color='k', lw=2, label=f'Среднее: {mean_log:.2f}')
ax.axvline(mean_log - 3 * std_log, color='r', lw=2, linestyle='--', label='-3σ')
ax.axvline(mean_log + 4 * std_log, color='r', lw=2, linestyle='--', label='+4σ')
ax.set_title('Распределение log(Возраст) с границами 3σ и 4σ', fontsize=13)
ax.set_xlabel('log(Возраст)')
ax.set_ylabel('Количество')
ax.legend()
plt.tight_layout()
plt.show()

# Границы: -3σ слева, +4σ справа
lower_bound = mean_log - 3 * std_log
upper_bound = mean_log + 4 * std_log

# Находим выбросы
mask_outliers = (log_age < lower_bound) | (log_age > upper_bound)

print(f"\nНижняя граница (-3σ): {lower_bound:.4f} → возраст {np.exp(lower_bound):.1f}")
print(f"Верхняя граница (+4σ): {upper_bound:.4f} → возраст {np.exp(upper_bound):.1f}")

print(f"\nНайдено выбросов: {mask_outliers.sum()}")
print("\nТаблица выбросов:")
print(df[mask_outliers][['Возраст', 'Опыт работы (лет)', 'ЗП (руб)']].sort_values('Возраст', ascending=False).head(20))

# Удаляем выбросы
df = df[~mask_outliers]

print(f"\nСтрок после очистки: {df.shape[0]}")
print(f"Удалено: {mask_outliers.sum()}")
```
